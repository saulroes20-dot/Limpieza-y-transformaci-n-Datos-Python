```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
from datetime import datetime
import openpyxl
```

```python
data1 = pd.read_csv('purchase_prices.csv', encoding='latin1')
data1.head(10)
```

```python
data = pd.read_csv('Ventas_Producto.csv', encoding='latin1')
data.head(10)
```

```python
#Inner Join (Solo Filas que coinciden en ambos DataFrames)

inner_join = pd.merge(data,data1, on='Brand')
print(inner_join)
```

```python
#LEFT JOIN (TODAS LAS FILAS DEL dATAFRAME IZQUIERDO )

left_join = pd.merge(data,data1, on='Brand', how='left')
print(left_join)
```

```python
#Right Join todo a la derecha 

left_join = pd.merge(data,data1, on='Brand', how='right')
print(left_join)
```

```python
#outer Join (todas las filas de ambos DataFrame)

left_join = pd.merge(data,data1, on='Brand', how='outer')
print(left_join)
```
