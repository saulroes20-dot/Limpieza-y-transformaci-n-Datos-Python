-----------------------Transformación de Datos--------

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
from datetime import datetime
import openpyxl
```

```python
data = pd.read_csv('Superstore.csv', encoding='latin1')
data.head(10)
```

```python
data.columns
```

```python
data.drop(columns='Ganacia Total ', inplace=True)
```

```python
# Aplicar una función a lo largo de un eje (filas o columnas), para saber el Máximo y el Mínimo, solamemte números

print(data.select_dtypes(include='number').apply(lambda x: x.max() - x.min()))
```

```python
#Solo sumame las filas númericas 
print(data.select_dtypes(include='number').sum(axis=1))
```

```python
data['Ganancia_Total'] = data[['Sales','Profit']].sum(axis=1)
```

```python
#Mostrar Solo filas númericas 
print(data.select_dtypes(include='number').sum(axis=1))
```

```python
#Transformar valores en una serie 
#Transformar valores x Columna
print(data['City'].map({'Los Angeles': 'L.A.' , 'Fort Worth ': 'FW'}))
```

```python
#Para agregar uan columna con los cambios 
data["City_Short"] = data['City'].map({'Los Angeles': 'L.A.' , 'Fort Worth ': 'FW'})
data2=data[['City_Short','City']]
print(data2)
```

```python
#Funciones con Map y en este caso vamor a sacar el 1.16 % Iva 

print(data['Sales'].map(lambda x: x*1.16))
```

```python
#Para agregar una nueva columna con el IVA 
data['+Iva'] = data['Sales'].map(lambda x: x*1.16)
```

```python
#applymap() aplicar una función a cada elemento del dataframe 
#Explicación, solamente aplica un ! a los valores int y str sino convierte en mayusculas str
print(data.map(lambda x: str(x) + '!' if isinstance(x, (int,float)) else x.upper()))
```

------------------Tranformaciones vectorizadas vs iterativas 

```python
#Método iterativo con el bucle for 
#cuenta el tiempo de cuanto tarda con el bucle for y te da la multiplicación entre dos columnas 
start = time.time()
resultado_for =[]

for i in range(len(data)):
    resultado_for.append(data.iloc[i]['Sales'] * data.iloc[i]['Profit'])
tiempo_for = time.time() - start
```

```python
print(tiempo_for)
```

```python
resultado_for
```

```python
#lo mismo pero con otra applicación apply 

start = time.time()
resultado_apply =[]

for i in range(len(data)):
    resultado_apply = data.apply(lambda x: x['Sales'] * x['Profit'], axis=1)
tiempo_for = time.time() - start
```

```python
#método más rápido 
start = time.time()
resultado_vectorizado = data['Sales'] * data['Profit']
tiempo_vectorizado = time.time() - start
```
