```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
from datetime import datetime
import openpyxl
```

```python
#Serie Strings 

s = pd.Series([' Hola Mundo  ', 'Python_para_Datos', 'pandas 1.3.0', 'CIENCIA de datos', 'email@ejemplo.com'])

print(s)


```

```python
print(s.str.lower()) #convierte todo a minusculas
print(s.str.upper()) #Convierte todo a mayusculas
print(s.str.strip()) #elimina espacios 
print(s.str.replace('_',' ')) #remplaza guión bajo por espacios 
```

```python
#Extracción de información 
print(s.str.len())
print(s.str.contains('datos', case=False)) #Case = False para ignorar mayusculas y minusculas
print(s.str[0:5])
```

```python
#Dividir string 
print(s.str.split(' '))
print(s.str.split(' ',expand=True))
```

```python
s = pd.Series([
    'Precio: 49.55 USD',
    'Costo: 123.50',
    'Sin Número',
    'Total 88.2 soles'
])
```

```python
#Expresiones Regulares 
print(s.str.extract(r'(\d+\.\d+)')) #extrae números con punto decimal
# la d sigbifica digitos
```

```python
s_emails = pd.Series([
    'correoexem@ejem.com',
    'imal@correo.com', 
    'otro.email@dominio.org'
    'no_es_email.com'])
```

```python
#Verifica si el strung es un email valido 
print(s_emails.str.match(r'^[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-zA-z]{2,}$'))
```

```python
data = pd.read_csv('Superstore.csv', encoding='latin1')
data.head(10)
```

```python
#Cambiar formato 
data['Fecha'] = pd.to_datetime(data['Order Date'])
data['Fecha_H'] = pd.to_datetime(data['Ship Date'])
```

```python
data.head(5)
```

```python
#Extraer componentes de fecha/hora
print(data['Fecha'].dt.year)
print(data['Fecha'].dt.month)
print(data['Fecha'].dt.day)
print(data['Fecha'].dt.day_name)
```

```python
print(data['Ship Date'].dt.hour)
print(data['Ship Date'].dt.minute)
print(data['Ship Date'].dt.second)
```

```python
#Sumar 7 días a las fecha

print(data['Fecha'] + pd.Timedelta(days=7))
```

```python
fecha_inicio = data['Fecha'].min()
fecha_final = data['Fecha'].max()
print(fecha_final-fecha_inicio)
```
